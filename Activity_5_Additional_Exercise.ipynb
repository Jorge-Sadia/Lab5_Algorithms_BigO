```python
# APPENDIX EXERCISE 1: EMERGENCY DISPATCH RECORD SEARCH
LAST_NAME = "Sadia"
STUDENT_ID = "TUPM-26-1950"
FAVORITE_ARTIST = "Taylor Swift"

SEED_DIGIT = int(STUDENT_ID[-1])  # 0
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())  # 23
ARTIST_SUM = sum(ord(c.upper()) - 64 for c in FAVORITE_ARTIST if c.isalpha()) # 153

# Data Generation Rules: Order maintained, unique values
base_records = [100 + ID_SUM + (i * 7) for i in range(15)]

# Rotation rule: Determine rotation point using SEED_DIGIT (Fallback to 4 since SEED_DIGIT is 0)
rotation_point = 4 if SEED_DIGIT == 0 else (SEED_DIGIT % len(base_records))
rotated_records = base_records[rotation_point:] + base_records[:rotation_point]

# Target Incident: Generate using ID_SUM and ARTIST_SUM (Wrapped to array length)
target_incident = base_records[(ID_SUM + ARTIST_SUM) % len(base_records)]

# Requirements: Search for target, record comparisons, compare with Linear Search
def rotated_binary_search(data, target):
    low, high, comps = 0, len(data) - 1, 0
    while low <= high:
        mid = (low + high) // 2
        comps += 1
        if data[mid] == target:
            return mid, comps
        if data[low] <= data[mid]:
            if data[low] <= target < data[mid]: high = mid - 1
            else: low = mid + 1
        else:
            if data[mid] < target <= data[high]: low = mid + 1
            else: high = mid - 1
    return -1, comps

def linear_search(data, target):
    comps = 0
    for i, val in enumerate(data):
        comps += 1
        if val == target: return i, comps
    return -1, comps

r_idx, r_comps = rotated_binary_search(rotated_records, target_incident)
_, l_comps = linear_search(rotated_records, target_incident)

print("=== ASSESSMENT DATA FOR EXERCISE 1 ===")
print("Original Dataset:", base_records)
print("Rotated dataset: ", rotated_records)
print("Target incident: ", target_incident)
print("Comparison Counts (Modified Binary):", r_comps)
print("Linear Search comparison count:     ", l_comps)
print("Time and space complexity: Time Complexity: O(log n) | Space Complexity: O(1)")
```

```python
# APPENDIX EXERCISE 2: UNIVERSITY ENROLLMENT RECORDS
LAST_NAME = "Sadia"
STUDENT_ID = "TUPM-26-1950"
FAVORITE_ARTIST = "Taylor Swift"

SEED_DIGIT = int(STUDENT_ID[-1])  # 0
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())  # 23
NAME_LENGTH = len(LAST_NAME)  # 5
ARTIST_LENGTH = len(FAVORITE_ARTIST)  # 12

# Data Generation Rules (Page 15): Ascending ordered datasets
dataset_A = sorted([10 + SEED_DIGIT + i * ID_SUM for i in range(5)])
dataset_B = sorted([5 + ARTIST_LENGTH + i * NAME_LENGTH for i in range(5)])

# Requirements: Consolidate two sorted datasets, count comparison operations
def merge_sorted_lists(list_a, list_b):
    merged, i, j, comps = [], 0, 0, 0
    while i < len(list_a) and j < len(list_b):
        comps += 1
        if list_a[i] <= list_b[j]:
            merged.append(list_a[i]); i += 1
        else:
            merged.append(list_b[j]); j += 1
    merged.extend(list_a[i:]); merged.extend(list_b[j:])
    return merged, comps

consolidated, total_comps = merge_sorted_lists(dataset_A, dataset_B)

print("=== ASSESSMENT DATA FOR EXERCISE 2 ===")
print("Department A dataset:", dataset_A)
print("Department B dataset:", dataset_B)
print("Consolidated dataset:", consolidated)
print(f"Number of records before and after consolidation: Before: {len(dataset_A)+len(dataset_B)} | After: {len(consolidated)}")
print("Comparison count:   ", total_comps)
print("Ordering verification (Is sorted?):", consolidated == sorted(consolidated))
print("Time and space complexity: Time Complexity: O(n + m) | Space Complexity: O(n + m)")
```

```python
# APPENDIX EXERCISE 3: E-COMMERCE ORDER PROCESSING
import random

STUDENT_ID = "TUPM-26-1950"
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())  

# Requirements (Page 16): Bubble, Selection, and Insertion Sort implementations
def bubble_sort_ex3(data):
    a, comps, moves = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comps += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]; moves += 1; swapped = True
        if not swapped: break
    return comps, moves

def selection_sort_ex3(data):
    a, comps, moves = data.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comps += 1
            if a[j] < a[minimum]: minimum = j
        if minimum != i: a[i], a[minimum] = a[minimum], a[i]; moves += 1
    return comps, moves

def insertion_sort_ex3(data):
    a, comps, moves = data.copy(), 0, 0
    for i in range(1, len(a)):
        key = a[i]; j = i - 1
        while j >= 0:
            comps += 1
            if a[j] <= key: break
            a[j + 1] = a[j]; moves += 1; j -= 1
        a[j + 1] = key
    return comps, moves

random.seed(ID_SUM)
base_orders = [random.randint(10, 99) for _ in range(15)]

# Three conditions: Sorted, reverse-ordered, and random
inputs = {
    "Sorted": sorted(base_orders),
    "Reverse": sorted(base_orders, reverse=True),
    "Random": base_orders.copy()
}

print("=== ASSESSMENT DATA FOR EXERCISE 3 ===")
print("Generated order dataset:", base_orders)
print("Sorted inputs:  ", inputs["Sorted"])
print("Reverse inputs: ", inputs["Reverse"])
print("Random inputs:  ", inputs["Random"])
print(f"\n{'Condition':<12}{'Algorithm':<16}{'Comparisons':>12}{'Movements':>12}")

for cond, dataset in inputs.items():
    for name, fn in [("Bubble Sort", bubble_sort_ex3), ("Selection Sort", selection_sort_ex3), ("Insertion Sort", insertion_sort_ex3)]:
        c, m = fn(dataset)
        print(f"{cond:<12}{name:<16}{c:>12}{m:>12}")
```
